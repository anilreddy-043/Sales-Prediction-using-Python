```python
# import the necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# step -2 : Load the dataset
df = pd.read_csv(r"C:\Users\anilr\OneDrive\Desktop\TASK 4 Sales Prediction\Advertising.csv")
```

```python
# Correlation Heatmap
# relationship between all numerical variables
plt.figure(figsize=(8, 5))

sns.heatmap(
    df.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# TV Advertising vs Sales
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='TV', y='Sales')
plt.title('TV Advertising vs Sales')
plt.xlabel('TV Advertising')
plt.ylabel('Sales')
plt.show()
```

```python
# Radio Advertising vs Sales
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Radio', y='Sales')
plt.title('Radio Advertising vs Sales')
plt.xlabel('Radio Advertising')
plt.ylabel('Sales')
plt.show()
```

```python
# Newspaper Advertising vs Sales
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Newspaper', y='Sales')
plt.title('Newspaper Advertising vs Sales')
plt.xlabel('Newspaper Advertising')
plt.ylabel('Sales')
plt.show()
```

```python
# Regression Plots
sns.regplot(data=df, x="TV", y="Sales")

plt.title("TV Advertising vs Sales with Regression Line")
plt.show()
```

```python
sns.regplot(data=df, x="Radio", y="Sales")

plt.title("Radio Advertising vs Sales with Regression Line")
plt.show()
```

```python
sns.regplot(data=df, x="Newspaper", y="Sales")

plt.title("Newspaper Advertising vs Sales with Regression Line")
plt.show()
```

```python
# checking for the outliers in the dataset using boxplots
plt.figure(figsize=(12, 6))
sns.boxplot(data=df)
plt.title("Boxplot for Outlier Detection")
plt.show()
```

```python
# Distribution Plots
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

sns.histplot(df["TV"], kde=True, ax=axes[0, 0])
axes[0, 0].set_title("TV Advertising Distribution")

sns.histplot(df["Radio"], kde=True, ax=axes[0, 1])
axes[0, 1].set_title("Radio Advertising Distribution")

sns.histplot(df["Newspaper"], kde=True, ax=axes[1, 0])
axes[1, 0].set_title("Newspaper Advertising Distribution")

sns.histplot(df["Sales"], kde=True, ax=axes[1, 1])
axes[1, 1].set_title("Sales Distribution")

plt.tight_layout()
plt.show()
```

```python
# Advertising Channel Comparison
total_spending = df[["TV", "Radio", "Newspaper"]].sum()

total_spending.plot(kind="bar", figsize=(7, 5))

plt.title("Total Advertising Spend by Channel")
plt.xlabel("Advertising Channel")
plt.ylabel("Total Spend")
plt.xticks(rotation=0)
plt.show()
```
